```python
import matplotlib.pyplot as plt

category_sales = df.groupby("category")["purchase_amount_(usd)"].sum()

plt.figure(figsize=(8,5))
category_sales.plot(kind="bar")

plt.title("Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Revenue (USD)")

plt.xticks(rotation=45)

plt.show()
```

```python
gender = df["gender"].value_counts()

plt.figure(figsize=(6,4))
gender.plot(kind="bar")

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")

plt.show()
```

```python
payment = df["payment_method"].value_counts()

plt.figure(figsize=(8,5))
payment.plot(kind="bar")

plt.title("Payment Method")

plt.xlabel("Payment")

plt.ylabel("Transactions")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["age"], bins=10)

plt.title("Age Distribution")

plt.xlabel("Age")

plt.ylabel("Customers")

plt.show()
```

```python
total_revenue = df["purchase_amount_(usd)"].sum()

print("Total Revenue (USD):", total_revenue)
```

```python
top_products = df.groupby("item_purchased")["purchase_amount_(usd)"] \
                 .sum() \
                 .sort_values(ascending=False) \
                 .head(10)

top_products
```

```python
plt.figure(figsize=(10,5))

top_products.plot(kind="bar")

plt.title("Top 10 Products by Revenue")
plt.xlabel("Product")
plt.ylabel("Revenue (USD)")
plt.xticks(rotation=45)

plt.show()
```

```python
payment_sales = df.groupby("payment_method")["purchase_amount_(usd)"] \
                  .sum() \
                  .sort_values(ascending=False)

payment_sales
```

```python
plt.figure(figsize=(8,5))

payment_sales.plot(kind="bar")

plt.title("Revenue by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Revenue (USD)")
plt.xticks(rotation=45)

plt.show()
```

```python
shipping_sales = df.groupby("shipping_type")["purchase_amount_(usd)"] \
                   .sum() \
                   .sort_values(ascending=False)

shipping_sales
```

```python
plt.figure(figsize=(8,5))

shipping_sales.plot(kind="bar")

plt.title("Revenue by Shipping Type")
plt.xlabel("Shipping Type")
plt.ylabel("Revenue (USD)")
plt.xticks(rotation=45)

plt.show()
```

```python
subscription_sales = df.groupby("subscription_status")["purchase_amount_(usd)"] \
                       .sum()

subscription_sales
```

```python
plt.figure(figsize=(5,4))

subscription_sales.plot(kind="bar")

plt.title("Revenue by Subscription Status")
plt.xlabel("Subscription")
plt.ylabel("Revenue (USD)")

plt.show()
```
